## **Atividade 01**

```python
import random
import json
import csv

# Gera 20 números aleatórios (inteiros entre 1 e 100)
numeros = [random.randint(1, 100) for _ in range(20)]

# Monta a lista de resultados com valor original, *2 e *10
dados = []
for valor in numeros:
    dados.append({
        "valor_original": valor,
        "valor_x2": valor * 2,
        "valor_x10": valor * 10
    })

# --- Salvar em TXT ---
with open("resultado.txt", "w") as f:
    f.write(f"{'Valor Original':<15}{'Valor x2':<15}{'Valor x10':<15}\n")
    for d in dados:
        f.write(f"{d['valor_original']:<15}{d['valor_x2']:<15}{d['valor_x10']:<15}\n")

# --- Salvar em JSON ---
with open("resultado.json", "w") as f:
    json.dump(dados, f, indent=4)

# --- Salvar em CSV ---
with open("resultado.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["valor_original", "valor_x2", "valor_x10"])
    writer.writeheader()
    writer.writerows(dados)

print("Arquivos gerados: resultado.txt, resultado.json, resultado.csv")
for d in dados:
    print(d)
```
