## **Exercício 01:**

### **Etapa 1** - Entrada e Cálculo

*   Criar três variáveis para armazenar as notas
*   Inicializar as variáveis com valores fixos;
*   Calcular a média das três notas;
*   Exibir as notas e a média;
*   Fazer o commit no GitHub, inserindo uma descrição da etapa.







```python
# Notas fixas
nota1 = 7.5
nota2 = 8.0
nota3 = 6.5

# Cálculo da média
media = (nota1 + nota2 + nota3) / 3

# Exibição dos resultados
print(f"Nota 1: {nota1}")
print(f"Nota 2: {nota2}")
print(f"Nota 3: {nota3}")
print(f"Média: {media:.2f}")
```

### **Etapa 2** - Melhorando o Programa

- Receber o nome do aluno usando `input()`
- Receber as três notas usando `input()`
- Converter as notas para `float`
- Calcular a média
- Exibir um pequeno relatório
- Fazer o commit no GitHub, inserindo uma descrição da etapa

```python
# Entrada de dados pelo usuário
nome = input("Digite o nome do aluno: ")
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))

# Cálculo da média
media = (nota1 + nota2 + nota3) / 3

# Relatório formatado
print("\n" + "=" * 25)
print("   RELATÓRIO DO ALUNO   ")
print("=" * 25)
print(f"Aluno: {nome}")
print(f"Notas: {nota1:.1f} | {nota2:.1f} | {nota3:.1f}")
print(f"Média Final: {media:.2f}")
print("=" * 25)
```
