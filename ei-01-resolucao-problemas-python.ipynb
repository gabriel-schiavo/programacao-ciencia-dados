## **Desafio 1** - Análise de notas de uma turma:
- Solicitar a quantidade de estudantes
- Armazenar as notas em uma lista
- Calcular (usando funções):
  - Maior nota
  - Menor nota
  - Média da turma
  - Quantidade de aprovados (nota ≥ 6,0)
  - Quantidade de reprovados

```python
# Funções de cálculo
def calc_maior(notas):
    return max(notas)

def calc_menor(notas):
    return min(notas)

def calc_media(notas):
    return sum(notas) / len(notas)

def num_aprovados(notas):
    return sum(1 for n in notas if n >= 6.0)

def num_reprovados(notas):
    return sum(1 for n in notas if n < 6.0)

# Entrada de dados
qtd = int(input("Quantidade de estudantes: "))
notas = []
for i in range(qtd):
    nota = float(input(f"Nota do aluno {i+1}: "))
    notas.append(nota)

# Exibição dos resultados
print("\n--- Resultados ---")
print(f"Maior nota: {calc_maior(notas):.1f}")
print(f"Menor nota: {calc_menor(notas):.1f}")
print(f"Média da turma: {calc_media(notas):.2f}")
print(f"Aprovados: {num_aprovados(notas)}")
print(f"Reprovados: {num_reprovados(notas)}")
```

## **Desafio 2** - Cadastro e consulta de produtos:
- Cadastrar produtos usando um dicionário, cada um com:
  - Código
  - Nome
  - Preço
  - Quantidade em estoque
- Permitir (usando funções):
  - Consultar um produto pelo código
  - Calcular o valor total do estoque
  - Identificar o produto de maior valor unitário

```python
# Funções do sistema
def cad_produtos():
    produtos = {}
    qtd = int(input("Quantos produtos deseja cadastrar? "))
    for _ in range(qtd):
        codigo = input("\nCódigo: ")
        nome = input("Nome: ")
        preco = float(input("Preço: R$ "))
        estoque = int(input("Quantidade em estoque: "))
        produtos[codigo] = {"nome": nome, "preco": preco, "qtd": estoque}
    return produtos

def consultar_produto(produtos, codigo):
    if codigo in produtos:
        p = produtos[codigo]
        print(f"Nome: {p['nome']} | Preço: R$ {p['preco']:.2f} | Estoque: {p['qtd']}")
    else:
        print("Produto não encontrado.")

def total_estoque(produtos):
    return sum(p["preco"] * p["qtd"] for p in produtos.values())

def mais_caro(produtos):
    return max(produtos.values(), key=lambda p: p["preco"])

# 1. Cadastro
produtos = cad_produtos()

# 2. Consulta
print("\n--- Consulta de Produto ---")
cod_busca = input("Digite o código do produto: ")
consultar_produto(produtos, cod_busca)


# 3. Relatórios
caro = mais_caro(produtos)
print("\n--- Resumo do Estoque ---")
print(f"Valor TOTAL em estoque: R$ {total_estoque(produtos):.2f}")
print(f"Produto de MAIOR valor UNITÁRIO: {caro['nome']} (R$ {caro['preco']:.2f}/un.)")
```
