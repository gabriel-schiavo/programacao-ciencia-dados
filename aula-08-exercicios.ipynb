## **Atividade 01:**

Usandos os métodos do Pandas, localize a sua cidade de residência e:
- mostre todas as informações
- qual o mês de temperatura mínima
- qual o mês de maior déficit hídrico
- a média de temperatura pelas 4 estações do ano

```python
import pandas as pd

#Importando um arquivo .csv dos dados de temperatura dos municípios
df = pd.read_csv('https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv')
```

```python
# Trabalhamos em uma cópia do dataframe original
df_ativ = df.copy()

# Filtra o dataframe para manter só a linha do município desejado
santa_rosa = df_ativ[(df_ativ['nome_municipio'] == 'Santa Rosa') & (df_ativ['uf'] == 'RS')]

# 1) Mostra todas as informações do município
display(santa_rosa)

# Lista com a sigla de cada mês, usada para montar os nomes das colunas mensais
meses = ['jan', 'fev', 'mar', 'abr', 'mai', 'jun', 'jul', 'ago', 'set', 'out', 'nov', 'dez']

# 2) Mês de temperatura mínima -> idxmin(axis=1) acha a coluna com menor valor, na linha
colunas_temp = ['temp_media_' + m for m in meses]
print('Mês de temperatura mínima:', santa_rosa[colunas_temp].idxmin(axis=1).values[0])

# 3) Mês de maior déficit hídrico -> idxmax(axis=1) acha a coluna com maior valor, na linha
colunas_def = ['def_' + m for m in meses]
print('Mês de maior déficit hídrico:', santa_rosa[colunas_def].idxmax(axis=1).values[0])

# 4) Média de temperatura por estação do ano (hemisfério sul)
# Em vez de repetir a mesma linha 4 vezes, guardamos os meses de cada estação
# em um dicionário e percorremos com um for
estacoes = {
    'Verão': ['dez', 'jan', 'fev'],
    'Outono': ['mar', 'abr', 'mai'],
    'Inverno': ['jun', 'jul', 'ago'],
    'Primavera': ['set', 'out', 'nov']
}

for nome_estacao, meses_estacao in estacoes.items():
    colunas_estacao = ['temp_media_' + m for m in meses_estacao]
    media = santa_rosa[colunas_estacao].mean(axis=1).values[0]
    print(f'{nome_estacao}: {media:.2f} °C')
```
