## **01** - Números primos:
  - Ler um número e verificar se é primo, testando o resto da divisão por 2 e pelos ímpares até o número lido.
  - Se algum resto for zero, o número não é primo.

  - **Desafio:** em vez de verificar um número específico, ler a *quantidade* de primos desejada e exibir essa quantidade de primeiros números primos (ex.: entrada 5 → lista dos 5 primeiros primos).



```python
num = int(input("Digite um número inteiro: "))

if num <= 1:
    primo = False
elif num == 2:
    primo = True
elif num % 2 == 0:
    primo = False
else:
    primo = True

    for divisor in range(3, num, 2):
        if num % divisor == 0:
            primo = False
            break

if primo:
    print(f"O número {num} é PRIMO.")
else:
    print(f"O número {num} NÃO é PRIMO.")
```

## **02** - Raiz quadrada pelo método de Newton:
  - Ler um número `n` e calcular sua raiz quadrada aproximada usando a fórmula `p = (b + (n/b)) / 2`.
  - A cada iteração, atualizar `b = p` e recalcular `p`.
  - Repetir até que a diferença absoluta entre `n` e o quadrado de `p` seja menor que 0,0001.
  - Ao final, calcular também a raiz usando a função `SQRT(n)` e exibir os dois resultados (o aproximado e o da função).

```python
import math

n = float(input("Digite um número: "))
b = n / 2

while True:
    p = (b + (n / b)) / 2
    if abs(n - (p ** 2)) < 0.0001:
        break
    b = p

print("Raiz calculada:", p)
print("Função sqrt:   ", math.sqrt(n))
```
