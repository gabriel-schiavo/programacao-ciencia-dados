## **Atividade 01:**
Construa uma calculadora básica usando os conceitos de orientação a objetos.

Use a função raiz quadrada usando o método de Newton (se achar complicado use a função SQRT).



```python
class Calculadora:
    # Construtor: recebe os dois números e a operação escolhida, e guarda tudo como atributos do objeto
    def __init__(self, a: float, b: float, operacao: str):
        self.a = a
        self.b = b
        self.operacao = operacao

    # Cada método abaixo representa uma operação matemática.
    # Eles recebem "a" e "b" como parâmetros, o que os torna reutilizáveis fora do contexto do objeto.
    def somar(self, a, b):
        return a + b

    def subtrair(self, a, b):
        return a - b

    def multiplicar(self, a, b):
        return a * b

    def dividir(self, a, b):
        # Trata divisão por zero antes de executar a conta
        if b == 0:
            raise ValueError("Divisão por zero não é permitida.")
        return a / b

    def raiz_quadrada(self, a, precisao=1e-10):
        # Não existe raiz real de número negativo
        if a < 0:
            raise ValueError("Não é possível calcular raiz de número negativo.")
        # Raiz de 0 é sempre 0
        if a == 0:
            return 0

        # Método de Newton para aproximar a raiz quadrada:
        # Começa o próprio número "a" como "chute" inicial
        x = a
        while True:
            # Fórmula de Newton: a cada passo, x se aproxima mais da raiz real
            proximo_x = 0.5 * (x + a / x)
            # Quando a diferença entre um valor e o próximo for muito pequena,
            # consideramos que coincide com a resposta
            if abs(proximo_x - x) < precisao:
                return proximo_x
            x = proximo_x  # atualiza x e repete o processo

    # Método para decidir qual operação executar
    def calcular(self):
        if self.operacao == "+":
            return self.somar(self.a, self.b)
        elif self.operacao == "-":
            return self.subtrair(self.a, self.b)
        elif self.operacao == "*":
            return self.multiplicar(self.a, self.b)
        elif self.operacao == "/":
            return self.dividir(self.a, self.b)
        elif self.operacao == "R":
            return self.raiz_quadrada(self.a)
        else:
            raise ValueError(f"Operação inválida: {self.operacao}")



# Conjunto com as operações aceitas
operacoes_validas = {"+", "-", "*", "/", "R"}

# entrada input() convertida para float
a = float(input("Digite o PRIMEIRO número: "))

operacao = input("Digite a operação (+, -, *, /, R para raiz): ").strip()

# Laço para continuar solicitando até o usuário digitar uma operação válida
while operacao not in operacoes_validas:
    operacao = input("Operação INVÁLIDA. Tente novamente (+, -, *, /, R): ").strip()

# Para raiz quadrada, o segundo número não é necessário,
# então é definido b = 0 automaticamente nesse caso
if operacao == "R":
    b = 0
else:
    b = float(input("Digite o SEGUNDO número: "))

# Cria o objeto Calculadora com os dados dos inputs
calc = Calculadora(a, b, operacao)

# Executa o cálculo com base na operação escolhida e mostra o resultado
print("\nRESULTADO:", calc.calcular())
```
