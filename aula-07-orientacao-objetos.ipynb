## **Classes:**

```python
# Definindo uma classe Pet e seus atribuitos
# Já estamos usando Abstração ao definir a classe
class Pet:
  # método construtor
  def __init__(self, nome: str, idade: int, peso: float, cor: str, tipo:str, __castrado=False):
    self.nome = nome
    self.idade = idade
    self.peso = peso
    self.cor = cor
    self.__tipo = tipo
    # a restrição de acesso aos atributos é definida ao inserir dois underscores ('_') ao atributo
    self.__castrado = __castrado

  def setTipo(self, tipo: str):
    self.__tipo = tipo

  def getTipo(self):
    return self.__tipo

  def setCor(self, cor: str):
    self.cor = cor

  def getCor(self):
    return self.cor

  def setCastracao(self):
    self.__anestesia()
    self.__castrado = True

  def getCastracao(self):
    return self.__castrado

  def apresentacao(self):
    print(f'Ola, meu nome e {self.nome}. Tenho {self.idade} anos, minha cor e {self.cor} e peso {self.peso} kg. ')

  # a restricão de acesso aos metodos e definida ao inserir dois underscores ('_') antes do nome
  def _anestesia(self):
    dose = self.peso / 1000

  # forca erro se não for implementado nas classes filhas
  def tipoComida(self):
    raise NotImplementedError()

  # para declarar um método sem conteudo (vazio) temos que usar o comando 'pass' (ja que o python nao tem limitadores de blocos)
  def altura(self):
    pass

  #0 repr_ serve para exibir o objeto para o programador, usada pelo console do Python
  def _repr_(self):
    return (
      f"{self.__class__.__name__}"
    )

  # O str_ serve para exibir o objeto para usuario final, usada pelo comando print e pela função str
  # def _str_(self):
    # return (
      # f"{self.__class__.__name__} - Olá, meu nome é {self.nome}. Tenho {self.idade} anos, minha cor é {self.cor} e peso {self.peso} kg.}"
    # )
```

### Usando a classe:

```python
# Instancia um objeto da Classe "Pet"
pet = Pet(nome='Ingo', idade=16, cor='Sal e Pimenta', peso=8.0, tipo='Pet')

# Chama os métodos de "Pet"
pet.apresentacao()
info = pet.__str__()
print(info)
print(pet.nome)
# Erro ao acessar o atributo
# print(pet.castrado)

# Uso do método
print(pet.getCastracao())
```

### Novo Construtor:

```python
# Definindo uma classe Pet e seus atribuitos
# Já estamos usando Abstração ao definir a classe

class Passaro(Pet):
  # método construtor
  def __init__(self, nome: str, idade: int, peso: float, cor: str, bico: str):
    super().__init__(nome, idade, peso, cor, 'Passaro')
    self.bico = bico

  def setBico(self, bico: str):
    self.bico = bico

  def getBico(self):
    return self.bico

  def tipoComida(self):
    return (
      "Alpiste"
    )
```

### Usando uma Classe Filha:

```python
cardeal = Passaro(nome='Pardal', idade=2, peso=0.7, cor='Amarela', bico='Fino')

print(cardeal.nome)
print(cardeal.bico)
print(cardeal.getCor())
print(cardeal.getTipo())
print(cardeal.tipoComida())
```
