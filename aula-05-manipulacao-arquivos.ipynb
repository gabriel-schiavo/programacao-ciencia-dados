## **TXT Files:**

### Abrir Arquivos

```python
# usando a funcão open para abrir um arquivo texto
# se não for especificado o modo de abertura do arquivo, por padrão é 'r' (leitura)
f = open("sample_data/README.md")

#pode ser necessario especificar o tipo de codificação do arquivo
#f = open("sample_data/README.md", mode='r', encoding='utf-8')

# aqui as operaçoes com o arquivo
conteudo = f.readlines()

#fechando o arquivo e liberando os recursos
f.close
```

### Visualizando o conteúdo do Arquivo

```python
# mostrando todo o conteúdo de uma vez
print (conteudo)

#quebrando em linhas o conteúdo
for linha in conteudo:
  print(linha)
```

### Gravando um Arquivo

```python
#manipulando o conteudo do arquivo
conteudo.reverse()

# abrindo um novo arquivo para gravação
arquivo_02 = open('sample_data/LEIAME.txt', 'w')

# escreve no disco o conteúdo da string
arquivo_02.writelines(conteudo)

# fecha o arquivo
arquivo_02.close()
```

#### Lendo conteúdo do arquivo LEIAME.txt:

```python
file_02 = open("sample_data/LEIAME.txt")

conteudo_02 = file_02.readlines()

print(conteudo_02)

for linha in conteudo_02:
  print(linha)
```

## **JSON Files:**

### JSON como String

```python
import json

#Podemos trabalhar o JSON como uma string
pessoa = '{"nome": "Pedro", "linguagens": ["Espanhol", "Português"]}'

# Ao fazer um parse em uma string JSON o python gera um dicionário
pessoa_dic = json.loads (pessoa)

# print: {'nome': 'Pedro', 'linguagens': ['Espanhol', 'Portugues']}
print(pessoa_dic)

# print: ['Espanhol', 'Portugues']
print(pessoa_dic['linguagens' ])
```

### JSON como Arquivo

```python
import json

with open('sample_data/anscombe.json', 'r') as f:
  dados = json.load(f)

#Print: { Series: ...}
print(dados)

# print: uma forma mais interessante de mostrar o JSON
print(json.dumps(dados, indent = 4, sort_keys=True))
```

### Criando arquivo com JSON

```python
#criando um dicionário com dados adicionais
pessoa_add = {"casado": True, "idade": 32}

# inserindo os dados adicionais no dicionario
pessoa_dic.update(pessoa_add)

#gravando ao arquivo formato json
with open('sample_data/pessoa.txt', 'w') as json_file:
  json.dump(pessoa_dic, json_file)
```

### JSON como API

```python
#importaro módulo json
import json
#importar o módulo requests para efetuar a requisição via API
import requests

#requisição via web para API
#resposta é um objeto - todas as possibilidades em https://requests.readthedocs.io/en/latest
resposta = requests.get("https://api.randomuser.me/")

#print se a requisição foi ok
print(resposta.status_code)

#print do texto da requisição
print(resposta.text)

#parse dos dados requisitados no formato json
dados = json.loads(resposta.text)

# print: dados brutos dos dicionário
print(dados)

# print: uma forma mais interessante de mostrar o JSON
print(json.dumps(dados, indent =4, sort_keys=True))
```

## **CSV Files:**

### Leitura de arquivo CSV

```python
#importa o módulo csv
import csv

# abre o arquivo com with
with open ('sample_data/california_housing_test.csv','r') as f:

  #funcao reader le o arquivo, de acordo com o delimitador espeficicado
  leitor = csv.reader(f,delimiter=',')
  for linha in leitor:
    total_quartos = linha[3]
    print(total_quartos)
```
