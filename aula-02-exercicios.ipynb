## **1. Conversão km/h → m/s**:
Programa que recebe a velocidade média em km/h e converte para m/s, dividindo pelo fator 3,6.

```python
v = float (input ("Informe uma velocidade média em km/h: "))

v_convertida = v / 3.6

print ("Velocidade média em m/s: ", v_convertida)
```

## **2. Distância entre dois pontos**:
Programa que recebe as coordenadas (x1, y1) do ponto P1 e (x2, y2) do ponto P2, calculando e exibindo a distância entre eles.

```python
from math import sqrt

x1 = int (input ("Informe o valor de X do P1: "))
y1 = int (input ("Informe o valor de Y do P1: "))

x2 = int (input ("Informe o valor de X do P2: "))
y2 = int (input ("Informe o valor de Y do P2: "))

result = sqrt (((x2 - x1)**2) + ((y2 - y1)**2))

print ("A distância entre P1 e P2 é: ", result)
```

## **3. Salário de corretor de imóveis**:
Programa que solicita nome do corretor, quantidade de imóveis vendidos e valor total das vendas, calculando o salário final com base em:
   - salário base de 2.500,00;
   - comissão fixa de 500,00 por imóvel vendido;
   - comissão de 5% sobre o valor total das vendas.

```python
nome = str (input ("Informe o NOME do corretor: "))
t_imoveis = int (input ("Informe o TOTAL de imóveis vendidos: "))
t_vendas = float (input ("Informe o VALOR TOTAL das vendas: "))


s_base = 2500
c_imoveis = 500 * t_imoveis
c_vendas = t_vendas * (5/100)
t_comissao = c_imoveis + c_vendas

result = s_base + t_comissao

print ("O SALÁRIO TOTAL do corretor", nome, "é:", result)
```
