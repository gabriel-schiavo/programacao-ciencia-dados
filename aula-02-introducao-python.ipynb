### **Programa Exemplo:**

```python
# Isso é um comentário (use # no inicio da frase)
# Cria a variável a e atribui o valor 5
a = 5

# Cria a variável b e atribui o valor 3.14
b = 3.14

# Cria a variavel c e atribui a mensagem "Programação"
c = "Programação"

# Printa na tela o valor da variável a e o tipo dela
print (a , type(a))
# Printa na tela o valor da variável b e o tipo dela
print (b, type(b))
# Printa na tela o valor da variável c e o tipo dela
print (c , type(c))

# Quais tipos foram definidos para as variáveis ?
```

### **Listas:**

```python
# Uma lista é uma sequência de valores que não necessitam ser do mesmo tipo

L = [2, 4, 5.98, "Programação", (4,7)]

print (L)

print (L[1])
print (L[4])
print (L[4] [1])
```

### **Tuplas:**

```python
# Na tupla os valores são imutáveis
# Lembre: Tupla usa parenteses ()

T = (10, "Programa", 200)

print (T[1])

# Desempacotamento de tupla
A, B, C = T

print (A)
print (B)
print (C)
```

### **Dicionários:**

```python
# No dicionario temos chave: valor
# Lembre: Dicionário usa chaves {}

D = { "Laranja": 10.50, "Banana" : 8.75, "Maça": 6.00}

print (D)
print (D["Laranja"])

D["Maca"] = 6.50

print (D)
```
