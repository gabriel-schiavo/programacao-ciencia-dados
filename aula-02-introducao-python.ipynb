### **Atividades:**

```python
# 1. Implemente um programa que converta o valor de uma velocidade média em km/h para m/s.
# Para isso, o usuário deve informar o valor da velocidade média. Sabe-se que o fator utilizado para essa conversão é 3,6.

v = float (input ("Informe uma velocidade média em km/h: "))

v_convertida = v / 3.6

print ("Velocidade média em m/s: ", v_convertida)
```

```python
# 2. Construa um programa que tem como dados de entrada dois pontos quaisquer no plano cartesiano: P1 e P2.
# Considere que P1 é definido pelas coordenadas x1 e y1, enquanto P2 por x2 e y2 .
# O programa deve calcular e escrever a distância entre os pontos P1 e P2.

from math import sqrt

x1 = int (input ("Informe o valor de X do P1: "))
y1 = int (input ("Informe o valor de Y do P1: "))

x2 = int (input ("Informe o valor de X do P2: "))
y2 = int (input ("Informe o valor de Y do P2: "))

result = sqrt (((x2 - x1)**2) + ((y2 - y1)**2))

print ("A distância entre P1 e P2 é: ", result)
```

```python
# 3. Uma imobiliária paga aos seus corretores um salário base de R$ 2.500,00.
# Além disso, uma comissão de R$ 500,00 por cada imóvel vendido e 5% do valor de cada venda.
# Construa um programa que solicite o nome do corretor, a quantidade de imóveis vendidos e o valor total de suas vendas.
# Ao fim, o programa deve calcular e escrever o salário final do corretor de imóveis.

nome = str (input ("Informe o NOME do corretor: "))
t_imoveis = int (input ("Informe o TOTAL de imóveis vendidos: "))
t_vendas = float (input ("Informe o VALOR TOTAL das vendas: "))


s_base = 2500
c_imoveis = 500 * t_imoveis
c_vendas = t_vendas * (5/100)
t_comissao = c_imoveis + c_vendas

result = s_base + t_comissao

print ("O SALÁRIO TOTAL do corretor", nome, "é:", result)
```
