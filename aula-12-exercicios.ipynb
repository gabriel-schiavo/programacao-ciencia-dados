## Atividade 01

O dataset não possui uma coluna de classificação, ela deve ser gerada, considerando a seguintes informações:
- Baixa: nota menor ou igual a 4.
- Média: nota 5 ou 6.
- Alta: nota maior ou igual a 7.

Obs. Essa classificação é para efeitos academicos

**Passos:**
1. Buscar o dataset
2. Inserir no pandas
3. Inserir uma coluna adicional com a classsificação ( de acorco com a tabela acima)
4. Separar os dados de características
5. Separar os dados de alvo (classificação)
6. Treinar o modelo KNN (escolha aleatoriamente o numero de vizinhos)
7. Confira a acurácia do algoritmo
8. Teste com outros valores de vizinhos buscando melhorar a acurácia.



### **1-2. Buscando e inserindo o Dataset no pandas**

```python
# uso da biblioteca pandas
import pandas as pd

url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv'

df = pd.read_csv(url, sep = ';')
print(f'\n {df}')
```

### **3. Inserindo a coluna de classificação**

```python
# função que transforma a nota em classificação
def classificar(nota):
  if nota <= 4:
    return 'Baixa'
  elif nota <= 6:
    return 'Media'
  else:
    return 'Alta'

# criando a nova coluna a partir da coluna quality
df['Classificacao'] = df['quality'].apply(classificar)
print(df['Classificacao'].value_counts())
print(f'\n {df[["quality", "Classificacao"]]}')
```

### **4-5. Separando os dados de características de alvo**

```python
# X será a matriz de características (sem quality e sem Classificacao, pois o alvo vem de quality)
x = df[df.columns.difference(['quality', 'Classificacao'])].values
print (f'X: {x}')

# y será o vetor com a classificação (target)
y = df['Classificacao'].values
print (f'Y: {y}')
```

### **6. Treinando o KNN**

```python
from sklearn.neighbors import KNeighborsClassifier

vinho_classificador = KNeighborsClassifier(n_neighbors = 3)
vinho_classificador.fit(x, y)
```

### **7. Conferindo a acurácia**

```python
from sklearn.model_selection import cross_val_score

# quebra o dataset em 5 partes e usa uma delas para teste (validação)
scores_dt = cross_val_score(vinho_classificador, x, y, scoring='accuracy', cv=5)

# pegamos o valor médio
print(f'Acuracia com 3 vizinhos: {scores_dt.mean()}')
```

### **8. Testando outros valores de vizinhos**

```python
# Treinando e avaliando - alterando o numero de vizinhos
for i in range(1,11):
  vinho_classificador = KNeighborsClassifier(n_neighbors = i)
  vinho_classificador.fit(x, y)
  scores_dt = cross_val_score(vinho_classificador, x, y, scoring='accuracy', cv=5)
  print(f'k: {i} -> Acuracia: {scores_dt.mean()}')
```
