## **Atividade I:**

Escreva um programa em Python que gere um número aleatório de 0 a 15, usando a biblioteca 'random' e o método 'randint()'. Na sequência, utilizando um loop faça com que o usuário tente acertar o número gerado.

A cada vez que o número é acertado, armazene num arquivo o número gerado e a quantidade de tentativas que o usuário realizou para acertar. Quando o usuário decidir parar de adivinhar, mostre todo o histórico que ele construiu durante a execução do programa.

```python
import random

ARQUIVO = 'atividade-01.txt'

# Função para gerar um número aleatório
def gerar_numero():
    return random.randint(0, 15)

# Função para salvar o hstórico da respectiva execução em um arquivo .txt
def salvar_historico(historico):
    with open(ARQUIVO, 'a', encoding='utf-8') as arquivo:
        for numero, tentativas in historico:
            arquivo.write(f'Número GERADO: {numero}\n')
            arquivo.write(f'Quantidade TENTATIVAS: {tentativas}\n')
            arquivo.write('\n-------------------------\n\n')

# Função para exibir o histórico da respectiva execução
def mostrar_historico(historico):
    print('===================')
    print('HISTÓRICO EXECUÇÃO:')
    print('===================\n')
    for numero, tentativas in historico:
        print(f'Número GERADO: {numero}')
        print(f'Quantidade TENTATIVAS: {tentativas}')
        print('\n-------------------------\n')

# Inicializa as váriaveis
numero = gerar_numero()
# print(numero)
tentativas = 0
historico = []

# Loop para execução do progrma até ser pressionado ENTER
while True:
    entrada = input("Digite um número no intervalo de 0-15 ou ENTER para ENCERRAR: ")

    # Encerra a execução caso ENTER seja pressionado no input
    if entrada == "":
        print('ENCERRADO!\n')
        salvar_historico(historico)
        mostrar_historico(historico)
        break

    num_digitado = int(entrada)

    tentativas += 1

    # Caso o número digitado seja igual ao numero gerado, salva o histórico e reinicia com novo número
    if num_digitado == numero:
        print('Você ACERTOU!')
        historico.append((numero, tentativas))
        numero = gerar_numero()
        # print(numero)
        tentativas = 0
    else:
        print('Você ERROU! Tente novamente.')
```

## **Atividade II:**

Crie um programa que permita adicionar itens para uma lista de compras.

O usuário vai adicionando a quantidade e o valor de cada item da lista. Ao final, deve mostrar o valor total das compras e salvar a lista num arquivo .CSV.

```python
import csv

ARQUIVO = 'atividade-02-lista-compras.csv'

itens = []

## Loop para cadastrar itens
while True:
    nome = input("Nome do item (ou ENTER para finalizar): ")

    # Encerra a execução caso ENTER seja pressionado no input
    if nome == "":
        break

    quantidade = int(input("Quantidade: "))
    valor = float(input("Valor unitário: R$ "))
    subtotal = quantidade * valor

    itens.append((nome, quantidade, valor, subtotal))

# Soma os valores de subtotal de todos os itens da lista
total = sum(item[3] for item in itens)

# Salva a lista de itens no arquivo .csv
with open(ARQUIVO, 'w', newline='', encoding='utf-8') as arquivo:
    escritor = csv.writer(arquivo)
    escritor.writerow(['Item', 'Quantidade', 'Valor Unitário', 'Subtotal'])
    escritor.writerows(itens)
    escritor.writerow(['TOTAL', '', '', total])

print(f"\nTotal da compra: R$ {total:.2f}")
print(f"Lista salva em '{ARQUIVO}'")

```
