## **IFs, Else, Elif:**


```python
nota = float (input("Digite o valor da nota: "))

if nota < 60:
  print ("Sua nota {} foi insuficiente! ". format(nota) )
```

```python
nota = float (input("Digite o valor da nota: "))

if nota < 60:
  print ("Sua nota {} foi insuficiente! ". format(nota) )
else:
  print ("Sua nota {} foi suficiente ".format(nota) )
```

```python
nota = float (input("Digite o valor da nota: "))

if nota < 60:
  print ("Sua nota {} foi insuficiente! ". format(nota) )
elif nota >= 60 and nota <=75:
  print ("Sua nota {} foi C! ". format(nota) )
elif nota > 75 and nota <=90:
  print ("Sua nota {} foi B! ".format(nota) )
else:
  print ("Sua nota {} A! ".format(nota) )
```

## **While:**

```python
# Uso do comando while para repeticao
senha = "112233"
digitado =" "

# Vai repetir o trecho de codigo enquanto a leitura foi diferente da senha
# Leitura: enquanto digitado e diferente de senha, repita
while (digitado != senha):
  digitado = input("Digite a senha: ")

  if digitado == senha :
    print('Acesso liberado ')
  else:
    print('Senha incorreta. Tente novamente')
```

## **For:**

```python
# For básico usando range
for i in range (1, 10, 1):
  print (i)
```

```python
# For básico usando range
for i in range (20, 10, -2):
  print (i)
```

```python
# For para passar pela lista
# Lista de notas de provas
notas_provas = [9, 6.5, 3.4, 8.5]
soma = 0

for nota in notas_provas:
  soma = soma + nota

média = soma / 4

# sem saber a quantidade de notas
# len(notas_provas)
print('Média = ', média)
```

```python
# Usamos o comando return para retornar valores de uma funcao e encerrar sua execucao
# Funcão para calcular percentual de um valor qualquer
# É possivel deixar um valor fixo para um argumento,
# quando ele não é informado na chamada, o valor padrão é utilizado
def calcula_percento(valor, percentual=100):
  quanto = valor * percentual / 100
  return quanto

print ('20% de 300 sao: {:10.2f}'.format (calcula_percento (300,20)))

print ('100% de 50 sao: {:10.2f}'.format (calcula_percento (50)))
```

## **Formatted Strings:**

```python
# Strings Formatadas use 'f' antes da mensagem
import math

# Uso de {} no local do valor
print(f'O valor de PI é: {math.pi:.4f}.')
```

```python
# Strings Formatadas usando metodo format
import math

# Uso de {} no local do valor
print('O valor de PI é: {:3.4f}.'.format(math.pi))
```
