# Criando um dataset importando um arquivo csv (não altere).

O arquivo *clima_historico_municipal.csv* contém um registro por município, identificado por código IBGE, nome e UF. Para cada município, a base informa o período de referência, a fonte climática, o método de processamento e um conjunto de variáveis mensais agregadas historicamente. As variáveis climáticas presentes são:

* tmin_*: temperatura mínima média mensal histórica
* tmax_*: temperatura máxima média mensal histórica
* temp_media_*: temperatura média mensal histórica, derivada de * tmin e tmax
* ppt_*: precipitação mensal histórica
* pet_*: evapotranspiração potencial mensal histórica
* aet_*: evapotranspiração real mensal histórica
* def_*: déficit hídrico mensal histórico

Cada sufixo mensal corresponde a um mês do ano:

*jan, fev, mar, abr, mai, jun, jul, ago, set, out, nov, dez*

Assim, por exemplo:

* temp_media_jan representa a média histórica de temperatura média para todos os janeiros do período processado
* ppt_mar representa a média histórica de precipitação para todos os meses de março do período processado
* def_set representa a média histórica de déficit hídrico para todos os meses de setembro do período processado

As unidades utilizadas são:
* temperaturas (*tmin, tmax, temp_media*): **graus Celsius (°C)**
* precipitação (*ppt*): **milímetros por mês (mm/mês)**
* evapotranspiração potencial (*pet*): **milímetros por mês (mm/mês)**
* evapotranspiração real (*aet*): **milímetros por mês (mm/mês)**
* déficit hídrico (*def*): **milímetros por mês (mm/mês)**

# Como renomear o nome das colunas

```python
'''
Comandos para renomear o nome das colunas do dataframe.
rename(columns={...}) -> renomeia colunas específicas, passando um dicionário
                          no formato {'nome_antigo': 'nome_novo'}
columns = [...] -> substitui TODOS os nomes das colunas de uma vez (lista completa)
Info: Gabriel
'''

# Sempre trabalhamos em uma cópia do dataframe original, para não alterar o df principal
df_renomeado = df.copy()

# Renomeia apenas as colunas indicadas, usando um dicionário {nome_atual: nome_novo}
df_renomeado = df_renomeado.rename(columns={
    'codigo_ibge': 'COD_IBGE',
    'nome_municipio': 'municipio',
    'uf': 'estado'
})

# Confirma que os nomes foram alterados
display(df_renomeado.head())
```

# Como adicionar linhas ao dataframe

```python
'''
Comandos para adicionar linhas ao dataframe.
pd.concat([df1, df2]) -> junta (empilha) dois dataframes, um embaixo do outro
ignore_index=True -> refaz o índice (0, 1, 2, ...) em vez de repetir índices antigos
Info: Gabriel
'''

# Trabalhamos em uma cópia do dataframe original
df_add = df.copy()

# Cria a nova linha como um dataframe de uma única linha.
# Não precisamos preencher todas as 91 colunas: as que não citarmos
# ficarão automaticamente com valor NaN (vazio) na linha nova.
nova_linha = pd.DataFrame([{
    'codigo_ibge': 9999999,
    'nome_municipio': 'Cidade Exemplo',
    'uf': 'XX',
    'periodo_inicio': 1975,
    'periodo_fim': 2024
}])

# pd.concat empilha os dois dataframes (df_add + nova_linha)
# ignore_index=True evita que o índice da nova linha se repita (fica 0 igual ao original)
df_add = pd.concat([df_add, nova_linha], ignore_index=True)

# Mostra as últimas linhas para conferir que a linha nova foi adicionada ao final
display(df_add.tail())
```

# Como adicionar colunas ao dataframe

```python
'''
Comandos para adicionar colunas ao dataframe.
df['nova_coluna'] = valor -> cria uma coluna nova (valor único ou calculado)
insert(posicao, nome, valores) -> cria a coluna em uma posição específica
Info: Gabriel
'''

# Trabalhamos em uma cópia do dataframe original
df_col = df.copy()

# Forma 1: coluna com um valor fixo, repetido em todas as linhas
df_col['pais'] = 'Brasil'

# Forma 2: coluna calculada a partir de outras colunas (operação é feita linha a linha)
# Aqui calculamos a amplitude térmica de janeiro (diferença entre máxima e mínima)
df_col['amplitude_termica_jan'] = df_col['tmax_jan'] - df_col['tmin_jan']

# Forma 3: insert() permite escolher a posição (índice) onde a coluna vai aparecer
# Aqui inserimos 'pais' como a primeira coluna (posição 0)
df_col.insert(0, 'pais_destaque', df_col['pais'])

# Confirma as novas colunas
display(df_col.head())
```

# Como deletar linhas

```python
'''
Comandos para deletar linhas do dataframe.
drop(index=[...]) -> remove linhas específicas, pelo número do índice
df[condição] -> forma alternativa: mantém só as linhas que atendem a condição
                (ou seja, 'deleta' as que não atendem)
Info: Gabriel
'''

# Trabalhamos em uma cópia do dataframe original
df_del = df.copy()

# Forma 1: dropar linhas pelo índice (aqui removemos as linhas 0, 1 e 2)
df_del = df_del.drop(index=[0, 1, 2])
print('Linhas após remover pelo índice:', df_del.shape[0])

# Forma 2: 'deletar' linhas mantendo apenas as que atendem a uma condição
# Aqui mantemos só os municípios que NÃO são de Rondônia (uf diferente de 'RO')
df_del = df_del[df_del['uf'] != 'RO']
print('Linhas após remover municípios de RO:', df_del.shape[0])

display(df_del.head())
```

# Como deletar colunas

```python
'''
Comandos para deletar colunas do dataframe.
drop(columns=[...]) -> remove uma ou mais colunas, pelo nome
drop([...], axis=1) -> forma equivalente, usando axis=1 para indicar 'colunas'
                        (axis=0 seria linhas, que é o padrão)
Info: Gabriel
'''

# Trabalhamos em uma cópia do dataframe original
df_col_del = df.copy()

# Forma 1: drop(columns=[...]) remove as colunas informadas pelo nome
df_col_del = df_col_del.drop(columns=['fonte_clima', 'metodo_processamento'])
print('Colunas após remover fonte_clima e metodo_processamento:', df_col_del.shape[1])

# Forma 2: mesmo resultado, usando axis=1 (1 = colunas, 0 = linhas)
df_col_del = df_col_del.drop(['periodo_inicio', 'periodo_fim'], axis=1)
print('Colunas após remover periodo_inicio e periodo_fim:', df_col_del.shape[1])

display(df_col_del.head())
```

# Como mesclar dois dataframe

```python
'''
Comandos para mesclar (juntar) dois dataframes.
pd.merge(df1, df2, on='coluna', how='left') -> junta as colunas de dois dataframes,
combinando as linhas que têm o mesmo valor na coluna informada em 'on'.
how define o tipo de junção: 'left', 'right', 'inner' ou 'outer'
Info: Gabriel
'''

# Trabalhamos em uma cópia do dataframe original
df_merge = df.copy()

# Cria um segundo dataframe pequeno, com a região de cada UF
df_regiao = pd.DataFrame({
    'uf': ['RO', 'GO', 'DF', 'SP', 'RS'],
    'regiao': ['Norte', 'Centro-Oeste', 'Centro-Oeste', 'Sudeste', 'Sul']
})

# pd.merge junta os dois dataframes usando a coluna 'uf' como referência
# how='left' mantém TODAS as linhas de df_merge, mesmo que a UF não esteja em df_regiao
# (nesse caso, a coluna 'regiao' fica como NaN para essas linhas)
df_merge = pd.merge(df_merge, df_regiao, on='uf', how='left')

# Mostra apenas algumas colunas para facilitar a conferência do resultado
display(df_merge[['nome_municipio', 'uf', 'regiao']].head())
```

# Buscar uma linha ou uma célula (linha e coluna) específica

```python
'''
Comandos para buscar uma linha ou uma célula específica (linha + coluna).
loc[]  -> busca pelo RÓTULO (nome) do índice e da coluna
iloc[] -> busca pela POSIÇÃO numérica (como se fosse uma lista, começando em 0)
Info: Gabriel
'''

# Trabalhamos em uma cópia do dataframe original
df_busca = df.copy()

# Busca a linha de índice/rótulo 10 (retorna uma Series com todas as colunas dessa linha)
linha_loc = df_busca.loc[10]
display(linha_loc)

# Busca a linha pela POSIÇÃO 10 (11ª linha do dataframe, contando do zero)
linha_iloc = df_busca.iloc[10]
display(linha_iloc)

# Busca uma célula específica: linha de índice 10, coluna 'nome_municipio' (por rótulo)
celula_loc = df_busca.loc[10, 'nome_municipio']
print('Município (loc):', celula_loc)

# Busca a mesma célula por POSIÇÃO: linha 10, coluna 1 (segunda coluna do dataframe)
celula_iloc = df_busca.iloc[10, 1]
print('Município (iloc):', celula_iloc)
```

# Buscar linhas de forma condicional

```python
'''
Comandos para buscar linhas de forma condicional (filtros).
df[condição] -> mantém apenas as linhas onde a condição é True
&  -> 'E' lógico (as duas condições precisam ser verdadeiras)
|  -> 'OU' lógico (basta uma das condições ser verdadeira)
isin([...]) -> verifica se o valor está dentro de uma lista de opções
IMPORTANTE: cada condição entre parênteses ao combinar com & ou |
Info: Gabriel
'''

# Trabalhamos em uma cópia do dataframe original
df_cond = df.copy()

# Condição simples: municípios do estado do Rio Grande do Sul
rs = df_cond[df_cond['uf'] == 'RS']
print('Municípios do RS:', rs.shape[0])

# Condição composta com & (E): municípios do RS que tiveram temp_media_jan acima de 24°C
rs_quente = df_cond[(df_cond['uf'] == 'RS') & (df_cond['temp_media_jan'] > 24)]
display(rs_quente[['nome_municipio', 'uf', 'temp_media_jan']].head())

# Condição composta com | (OU): municípios do RS OU de SC
sul = df_cond[(df_cond['uf'] == 'RS') | (df_cond['uf'] == 'SC')]
print('Municípios do RS ou SC:', sul.shape[0])

# isin() é uma forma mais prática de comparar com VÁRIOS valores ao mesmo tempo
sul_isin = df_cond[df_cond['uf'].isin(['RS', 'SC', 'PR'])]
print('Municípios da região Sul (RS, SC, PR):', sul_isin.shape[0])
```

# Buscar um dataframe parcial (Linhas e Colunas específicas)

```python
'''
Comandos para buscar um dataframe PARCIAL (linhas e colunas específicas ao mesmo tempo).
loc[linhas, colunas]  -> seleciona por RÓTULO (nome do índice / nome das colunas)
iloc[linhas, colunas] -> seleciona por POSIÇÃO (números, começando em 0)
Info: Gabriel
'''

# Trabalhamos em uma cópia do dataframe original
df_parcial = df.copy()

# loc: linhas de índice 0 a 5 (inclusive) e apenas as colunas informadas pelo nome
parcial_loc = df_parcial.loc[0:5, ['nome_municipio', 'uf', 'temp_media_jan', 'ppt_jan']]
display(parcial_loc)

# iloc: linhas nas posições 0 a 5 (0:6, o 6 não entra) e colunas nas posições 0 a 2
parcial_iloc = df_parcial.iloc[0:6, 0:3]
display(parcial_iloc)
```

# Como agrupar valores de uma coluna. Semelhante a banco de dados: Groupby


```python
'''
Comandos para agrupar valores de uma coluna (groupby), parecido com GROUP BY do SQL.
groupby('coluna') -> agrupa as linhas que têm o mesmo valor nessa coluna
Depois do groupby, aplicamos uma função de agregação: mean(), sum(), count(), etc.
agg({...}) -> permite aplicar funções diferentes para colunas diferentes de uma vez
Info: Gabriel
'''

# Trabalhamos em uma cópia do dataframe original
df_group = df.copy()

# Agrupa por UF e calcula a média da temperatura média de janeiro em cada estado
media_temp_uf = df_group.groupby('uf')['temp_media_jan'].mean()
display(media_temp_uf)

# Agrupa por UF e conta quantos municípios existem em cada estado
qtd_municipios_uf = df_group.groupby('uf')['nome_municipio'].count()
display(qtd_municipios_uf)

# agg() permite aplicar VÁRIAS funções/colunas diferentes na mesma linha de código
resumo_uf = df_group.groupby('uf').agg({
    'temp_media_jan': 'mean',   # média da temperatura de janeiro
    'ppt_jan': 'sum',           # soma da precipitação de janeiro
    'nome_municipio': 'count'   # quantidade de municípios
})
display(resumo_uf.head())
```

# Criar um dataframe a partir de um dicionário

```python
'''
Como criar um dataframe a partir de um dicionário (sem precisar de um arquivo csv).
Forma 1: dicionário onde cada CHAVE vira uma coluna e cada VALOR é uma lista com os dados
Forma 2: lista de dicionários, onde cada dicionário representa uma LINHA
Info: Gabriel
'''

# Forma 1: dicionário de listas -> cada chave é uma coluna, cada lista são os valores
# Todas as listas precisam ter o MESMO tamanho (mesma quantidade de linhas)
dados_uf = {
    'uf': ['RS', 'SC', 'PR'],
    'capital': ['Porto Alegre', 'Florianópolis', 'Curitiba'],
    'regiao': ['Sul', 'Sul', 'Sul']
}
df_uf = pd.DataFrame(dados_uf)
display(df_uf)

# Forma 2: lista de dicionários -> cada dicionário vira uma linha do dataframe
dados_linhas = [
    {'uf': 'RS', 'capital': 'Porto Alegre', 'regiao': 'Sul'},
    {'uf': 'SC', 'capital': 'Florianópolis', 'regiao': 'Sul'},
    {'uf': 'PR', 'capital': 'Curitiba', 'regiao': 'Sul'}
]
df_uf_linhas = pd.DataFrame(dados_linhas)
display(df_uf_linhas)
```

# Como deletar linhas com valores vazios

```python
'''
Comandos para deletar linhas com valores vazios (NaN).
dropna() -> remove as linhas que têm PELO MENOS UM valor vazio
subset=[...] -> considera valores vazios apenas nas colunas informadas
how='all' -> só remove a linha se TODAS as colunas estiverem vazias
Info: Gabriel
'''

# Trabalhamos em uma cópia do dataframe original
df_na = df.copy()

# O dataset original não tem valores vazios, então criamos alguns artificialmente
# apenas para demonstrar o funcionamento do dropna()
df_na.loc[0, 'temp_media_jan'] = None
df_na.loc[1, 'ppt_jan'] = None
print('Linhas antes de remover valores vazios:', df_na.shape[0])

# dropna() remove QUALQUER linha que tenha pelo menos uma célula vazia
df_sem_na = df_na.dropna()
print('Linhas após dropna() padrão:', df_sem_na.shape[0])

# subset=[...] limita a verificação a colunas específicas
# aqui só removemos a linha se 'temp_media_jan' estiver vazia (ignora as demais colunas)
df_sem_na_subset = df_na.dropna(subset=['temp_media_jan'])
print('Linhas após dropna(subset=["temp_media_jan"]):', df_sem_na_subset.shape[0])
```

# Como preencher valores vazios

```python
'''
Comandos para preencher valores vazios (NaN), em vez de simplesmente apagá-los.
fillna(valor) -> substitui os valores vazios por um valor fixo
fillna(df['coluna'].mean()) -> substitui pela média da própria coluna
ffill() -> preenche com o valor da linha ANTERIOR (forward fill)
bfill() -> preenche com o valor da linha SEGUINTE (backward fill)
Info: Gabriel
'''

# Trabalhamos em uma cópia do dataframe original
df_fill = df.copy()

# O dataset original não tem valores vazios, então criamos alguns artificialmente
df_fill.loc[0, 'temp_media_jan'] = None
df_fill.loc[2, 'temp_media_jan'] = None

# Forma 1: preencher com um valor fixo (ex: zero)
df_fill_zero = df_fill['temp_media_jan'].fillna(0)
print('Preenchido com 0:', df_fill_zero.head(3).tolist())

# Forma 2: preencher com a MÉDIA da própria coluna (mais comum em dados numéricos)
media_col = df_fill['temp_media_jan'].mean()
df_fill_media = df_fill['temp_media_jan'].fillna(media_col)
print('Preenchido com a média (%.2f):' % media_col, df_fill_media.head(3).tolist())

# Forma 3: ffill() repete o último valor válido visto antes da lacuna
df_fill_ffill = df_fill['temp_media_jan'].ffill()
print('Preenchido com ffill:', df_fill_ffill.head(3).tolist())
```

# Comandos para inserir colunas de peso e altura no sistema universal

```python
'''
Comandos para inserir colunas de peso e altura no sistema universal (métrico).
Aqui simulamos um pequeno dataframe com peso em libras (lb) e altura em polegadas (in),
unidades do sistema imperial, e criamos colunas novas convertidas para o sistema
métrico (universal): quilogramas (kg) e centímetros (cm).
1 libra (lb)   = 0.453592 kg
1 polegada (in) = 2.54 cm
Info: Gabriel
'''

# Dataframe de exemplo, criado a partir de um dicionário (visto na célula anterior)
dados_pessoas = {
    'nome': ['Ana', 'Bruno', 'Carla'],
    'peso_lb': [130, 180, 150],     # peso em libras
    'altura_in': [64, 70, 66]       # altura em polegadas
}
df_pessoas = pd.DataFrame(dados_pessoas)

# Cria a coluna de peso no sistema universal (kg), convertendo a partir de libras
df_pessoas['peso_kg'] = df_pessoas['peso_lb'] * 0.453592

# Cria a coluna de altura no sistema universal (cm), convertendo a partir de polegadas
df_pessoas['altura_cm'] = df_pessoas['altura_in'] * 2.54

# Arredonda os valores convertidos para 1 casa decimal, só para facilitar a leitura
df_pessoas['peso_kg'] = df_pessoas['peso_kg'].round(1)
df_pessoas['altura_cm'] = df_pessoas['altura_cm'].round(1)

display(df_pessoas)
```
