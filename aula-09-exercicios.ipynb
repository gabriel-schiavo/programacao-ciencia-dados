## **Atividade 01:**
Escrever um programa usando Selenium para Scrap da:

* temperatura (mínima e máxima) do dia;
* de uma cidade qualquer (informada pelo usuário);
 * de algum site de previsão do tempo.

```python
# Instalao Selenium
!pip install -q google-colab-selenium
```

```python
# Importações
import google_colab_selenium as gs
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from time import sleep

# Configura o navegador
navegador = gs.Chrome()

# Le a cidade digitada pelo usuário
cidade = input("Digite o nome de uma cidade no RS: ")

# Monta o endereço de busca do clima
url = (
    "https://tempo.clicrbs.com.br/rs/"
    f"{cidade.replace(' ', '-')}/"
)

# Acessa o endereço de clima
navegador.get(url)
sleep(2)
print("Página:", navegador.title)
sleep(2)

# Obtém a data da previsão
data = navegador.find_element(By.XPATH, "/html/body/section[3]/div[2]/div/ol/li[1]/a/div[1]/h4").text

# Obtém a temperatura atual
temp_atual = navegador.find_element(By.XPATH, "/html/body/section[3]/div[1]/div/div[1]/div[1]/div[1]/span").text

# Obtém a temperatura máxima do dia
temp_max = navegador.find_element(By.XPATH, "/html/body/section[3]/div[2]/div/ol/li[1]/a/div[2]/div[2]/span[1]").text

# Obtém a temperatura mínima do dia
temp_min = navegador.find_element(By.XPATH, "/html/body/section[3]/div[2]/div/ol/li[1]/a/div[2]/div[2]/span[2]").text

# Exibe os dados coletados
print("Dia:", data)
print("Temperatura atual:", temp_atual)
print("Máxima do dia:", temp_max)
print("Mínima do dia:", temp_min)

# Finaliza o navegador
navegador.quit()
```
